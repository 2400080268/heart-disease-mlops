# Exploratory analysis: UCI Cleveland Heart Disease

Run `python scripts/run_pipeline.py` from the project root first. This notebook gives a small visual overview; the reproducible training pipeline remains in the scripts. The target is mapped to 0 (no disease) and 1 (disease).

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd() if (Path.cwd() / 'data').exists() else Path.cwd().parent
data = pd.read_csv(ROOT / 'data/processed/heart_disease.csv')
data.head()
```

```python
print('Rows and columns:', data.shape)
print('Missing values by column:')
display(data.isna().sum())
print('Duplicate rows:', data.duplicated().sum())
display(data.describe(include='all').T)
```

```python
counts = data['target'].value_counts().sort_index()
ax = counts.rename(index={0: 'No disease', 1: 'Disease'}).plot.bar(color=['#3977a8', '#d6843c'])
ax.set(title='Target class counts', xlabel='Class', ylabel='Rows')
plt.tight_layout()
plt.show()
```

```python
correlation = data.select_dtypes(include='number').corr()
fig, ax = plt.subplots(figsize=(10, 8))
image = ax.imshow(correlation, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
ax.set_yticks(range(len(correlation.columns)), correlation.columns)
ax.set_title('Numeric feature and target correlations (exploratory)')
fig.colorbar(image, ax=ax, label='Correlation')
fig.tight_layout()
plt.show()
```

Review missingness and class balance before modeling. Numeric correlations are exploratory only; they do not imply causal relationships. The dataset is small and historical, and the model is not a medical diagnostic system.
